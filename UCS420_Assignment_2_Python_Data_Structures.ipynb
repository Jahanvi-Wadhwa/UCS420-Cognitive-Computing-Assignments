## Question 1 — Lists

Take the roll number, extract its digits, and multiply each digit by 10.

```python
# Q1: Create list L from roll-number digits

digits = [int(digit) for digit in str(roll_number)]
L = [digit * 10 for digit in digits]

print("Original digits:", digits)
print("L:", L)

# ii. Add two numbers: append() and insert()
L.append(50)          # append() adds the value at the end
print("After append(50):", L)

L.insert(1, 70)        # insert() adds 70 at index 1
print("After insert(1, 70):", L)

# iii. Remove two elements: remove() and pop()
L.remove(50)          # remove() removes the first occurrence of value 50
print("After remove(50):", L)

removed = L.pop(1)    # pop() removes and returns the element at index 1
print("Popped element:", removed)
print("After pop(1):", L)

# iv. Sort ascending and descending
L.sort()
print("Ascending order:", L)

L.sort(reverse=True)
print("Descending order:", L)

# v. Slicing
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])

# vi. Elements greater than average
average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]

print("Average of L:", average)
print("Elements greater than average:", greater_than_average)
```

## Question 2 — Tuples

The first 8 values from the original digit×10 list are used to create `scores`.

```python
# Q2: Create tuple using the first 8 values from the original L
scores = tuple(digit * 10 for digit in digits[:8])

print("scores:", scores)

# i. Highest score and its index; lowest score and its frequency
highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)

# ii. Reverse tuple and return as a list
# Tuples are immutable, so they cannot be reversed in place.
reversed_scores = list(reversed(scores))
print("Reversed tuple as list:", reversed_scores)

# iii. Search for a score
search_score = int(input("Enter a score to search in scores: "))

if search_score in scores:
    print("First occurrence index:", scores.index(search_score))
else:
    print("Score is not present in the tuple.")

# iv. Attempt to change a tuple element
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
    print("Tuples are immutable, unlike lists, so their elements cannot be changed directly.")

# v. Tuple unpacking using *
first_score, second_score, *remaining_scores = scores
print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

## Question 3 — Random Numbers

The random seed is set to the user's roll number so the same roll number always produces the same random list.

```python
# Q3: Generate 100 reproducible random numbers

import random
from collections import Counter

random.seed(roll_number)

random_numbers = [random.randint(100, 900) for _ in range(100)]

print("Random list:")
print(random_numbers)

# ii. Count and print all odd numbers
odd_numbers = [x for x in random_numbers if x % 2 != 0]
print("\nOdd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))

# iii. Count and print all even numbers
even_numbers = [x for x in random_numbers if x % 2 == 0]
print("\nEven numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))

# iv. Prime numbers using list comprehension
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in random_numbers if is_prime(x)]

print("\nPrime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))

# v. Most frequent number and its frequency
frequency = Counter(random_numbers)
most_frequent_number, frequency_count = frequency.most_common(1)[0]

print("\nMost frequent number:", most_frequent_number)
print("Frequency:", frequency_count)
```

## Question 4 — Sets

Create sets A and B using the original 8 roll-number digits:
- A = digit × 7
- B = digit × 9

```python
# Q4: Create sets A and B

first_8_digits = digits[:8]

A = {digit * 7 for digit in first_8_digits}
B = {digit * 9 for digit in first_8_digits}

print("Set A:", A)
print("Set B:", B)

# vi. Union
union = A.union(B)
print("\nUnion of A and B:", union)

# vii. Intersection
intersection = A.intersection(B)
print("Intersection of A and B:", intersection)

# viii. Difference
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() keeps only elements belonging to the first set
# symmetric_difference() keeps elements belonging to either set but not both.

# ix. Symmetric difference
symmetric_difference = A.symmetric_difference(B)
print("Symmetric difference:", symmetric_difference)

# x. Subset and superset checks
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# xi. discard()
X = int(input("\nEnter a value to remove from set A: "))
A.discard(X)

print("Set A after discard:", A)
# discard() is safer than remove() because it does not raise an error
# when the value is not present in the set.
```

## Question 5 — Dictionaries

Create the given dictionary using the user's details.

```python
# Q5: Create the original dictionary

my_dict = {
    "name": name,
    "roll_no": roll_number,
    "branch": branch,
    "age": age,
    "city": city
}

print("Original dictionary:")
print(my_dict)

# i. Rename "city" to "location" without recreating the dictionary
my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming city to location:")
print(my_dict)

# ii. Add CGPA
my_dict["cgpa"] = cgpa
print("\nAfter adding CGPA:")
print(my_dict)

# iii. Increase age by 1
my_dict["age"] += 1
print("\nAfter increasing age by 1:")
print(my_dict)

# iv. Delete branch using pop() and del in separate copies

dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")

dict_del = my_dict.copy()
del dict_del["branch"]

print("\nDictionary after pop():")
print(dict_pop)
print("Value returned by pop():", removed_branch)

print("\nDictionary after del:")
print(dict_del)

# pop() removes the key and returns its value;
# del removes the key but does not return its value.

# v. Iterate using items()
print("\nKey-value pairs:")
for key, value in my_dict.items():
    print(f"{key} → {value}")

# vi. Check for email safely
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key does not exist.")

# vii. Create friend_dict and merge
friend_dict = {
    "name": "Aarav",
    "roll_no": 999,
    "branch": branch,
    "age": age,
    "city": "Jaipur"
}

# The assignment asks for the same 5 original keys.
merged_dict = {**my_dict, **friend_dict}

print("\nFriend dictionary:")
print(friend_dict)

print("\nMerged dictionary:")
print(merged_dict)

# When the same key exists in both dictionaries,
# the value from the second dictionary (friend_dict) wins.

# viii. Dictionary comprehension for string values
string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nOnly key-value pairs whose values are strings:")
print(string_values)
```
