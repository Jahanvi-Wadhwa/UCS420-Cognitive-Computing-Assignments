## Q1. Sensor readings: vectorization and Boolean indexing

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# (a) Add the average measurement error of +2°C to every reading
corrected_temperature = temperature + 2
print("Corrected temperatures (°C):", corrected_temperature)

# (b) Convert corrected temperatures from Celsius to Fahrenheit
fahrenheit = (9 / 5) * corrected_temperature + 32
print("Temperatures (°F):", fahrenheit)

# (c) Identify corrected readings greater than 32°C using Boolean indexing
above_32 = corrected_temperature[corrected_temperature > 32]
print("Corrected readings above 32°C:", above_32)

# (d) Count readings exceeding 32°C
count_above_32 = np.count_nonzero(corrected_temperature > 32)
print("Number of readings above 32°C:", count_above_32)
```

**(e) Explanation:** Vectorization applies an operation to an entire NumPy array without writing an explicit Python `for` loop. NumPy performs these operations efficiently on array data. Boolean indexing creates a True/False condition and selects only the values that satisfy it, making filtering concise and efficient.

## Q2. Daily step counts

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# (a) Total steps across all users and days
print("Total steps:", np.sum(steps))

# (b) Mean number of steps
print("Mean steps:", np.mean(steps))

# (c) Maximum and minimum values
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

# (d) Total steps for each day (sum down the rows)
print("Total steps for each day (axis=0):", np.sum(steps, axis=0))

# (e) Total steps for each user (sum across columns)
print("Total steps for each user (axis=1):", np.sum(steps, axis=1))

# (f) Position of the maximum value
flat_index = np.argmax(steps)
user_index, day_index = np.unravel_index(flat_index, steps.shape)
print("Maximum value position (zero-based user, day):", (user_index, day_index))
print("Maximum value position (one-based user, day):", (user_index + 1, day_index + 1))
```

## Q3. Slicing, copying, reshaping, flattening, and array properties

```python
# (a) Create the original array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

# (b) Slice indices 1 through 4 (index 4 is excluded)
subset = original[1:4]
print("\nSubset:", subset)

# (c) A basic slice is a view, so modifying it also changes original
subset[0] = 999
print("After modifying subset:")
print("Original:", original)
print("Subset:", subset)

# (d) .copy() creates an independent array
copied_subset = original[1:4].copy()
copied_subset[0] = 500
print("\nAfter modifying copied subset:")
print("Original:", original)
print("Copied array:", copied_subset)

# (e) Create values 1 through 12 and reshape to 3 × 4
matrix = np.arange(1, 13).reshape(3, 4)
print("\n3 × 4 matrix:\n", matrix)

# (f) Requested indexing and slicing
print("\nFirst row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
# Rows 1–2 and columns 2–3, interpreted as one-based positions
print("Rows 1–2 and columns 2–3:\n", matrix[0:2, 1:3])

# (g) flatten() and ravel()
flattened = matrix.flatten()
raveled = matrix.ravel()
print("\nflatten():", flattened)
print("ravel():", raveled)

# (h) ravel() usually returns a view for this contiguous matrix
raveled[0] = 1000
print("\nAfter changing ravel result:")
print("raveled:", raveled)
print("Matrix (affected):\n", matrix)

# (i) flatten() returns a copy; changing it does not affect matrix
flattened[1] = 2000
print("\nAfter changing flatten result:")
print("flattened:", flattened)
print("Matrix (not affected by flatten change):\n", matrix)

# (j) Matrix properties
print("\nShape:", matrix.shape)
print("Dimensions (ndim):", matrix.ndim)
print("Number of elements (size):", matrix.size)
print("Data type (dtype):", matrix.dtype)
```

**Q3 note:** Python slicing excludes the stop index, so `original[1:4]` contains indices 1, 2, and 3. A basic slice is a view; `.copy()` creates independent data. For a contiguous array, `ravel()` can return a view, whereas `flatten()` always returns a copy.

## Q4. Ordinary Least Squares (OLS) for assistance score

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
], dtype=float)

y = np.array([40, 65, 30, 85], dtype=float)

# (a) Shape and dimensions
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# (b) Transpose: rows become columns and columns become rows
print("\nX transpose (X.T):\n", X.T)

# (c) Matrix product X.T @ X
XtX = X.T @ X
print("\nX.T @ X:\n", XtX)

# (d) Inverse, if the matrix is invertible
XtX_inverse = np.linalg.inv(XtX)
print("\nInverse of X.T @ X:\n", XtX_inverse)

# (e) OLS coefficients: beta = (X.T X)^(-1) X.T y
beta = XtX_inverse @ X.T @ y
print("\nOLS coefficients (no intercept term):", beta)

# (f) Interpret coefficients
print("\nCoefficient order: sleep hours, activity level, stress level.")
print("Each coefficient is the model's estimated change in score for a one-unit")
print("increase in that feature, holding the other features constant.")

# (g) Predict score for a new user
new_user = np.array([5, 40, 7], dtype=float)
predicted_score = new_user @ beta
print("\nPredicted assistance score:", predicted_score)
```

**Q4 interpretation note:** The assignment's equation uses the three feature columns directly and does not include a column of ones, so this solution fits a model **without an intercept**. The coefficients are estimates based on only four sample users; treat the prediction as a mathematical exercise rather than a validated real-world score.
