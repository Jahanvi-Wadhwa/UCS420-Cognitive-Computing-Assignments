## Q1. Create a 1-D array and perform basic operations

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)
print("Addition of 2:", arr + 2)
print("Multiplication by 3:", arr * 3)
print("Division by 2:", arr / 2)
```

## Q2(b). Find the most frequent value and its index/indices

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def most_frequent_and_indices(array):
    values, counts = np.unique(array, return_counts=True)
    max_count = counts.max()
    most_frequent_values = values[counts == max_count]
    print("Array:", array)
    print("Most frequent value(s):", most_frequent_values)
    for value in most_frequent_values:
        print(f"Indices of {value}:", np.where(array == value)[0])
    print()

most_frequent_and_indices(x)
most_frequent_and_indices(y)
```

## Q3. Access elements in a 2-D array

## Q4. Create an array with 25 evenly spaced numbers from 10 to 100

`ndim` gives the number of dimensions, `shape` gives the size of each dimension, `size` gives the total number of elements, `dtype` gives the element data type, and `nbytes` gives the total memory used by the array.

```python
Jahanvi = np.linspace(10, 100, 25)

print("Array:", Jahanvi)
print("Dimensions:", Jahanvi.ndim)
print("Shape:", Jahanvi.shape)
print("Total elements:", Jahanvi.size)
print("Data type:", Jahanvi.dtype)
print("Total bytes consumed:", Jahanvi.nbytes)

# A 1-D array can be reshaped into a 2-D array (5 rows × 5 columns)
Jahanvi_transpose = Jahanvi.reshape(5, 5).T
print("\nTranspose after reshaping to 5 × 5:")
print(Jahanvi_transpose)

# Can we use T directly?
print("\nUsing .T on the original 1-D array:")
print(Jahanvi.T)
print("For a 1-D array, .T does not change its shape. Reshape to 2-D first to transpose.")
```

## Q5. Create a 3 × 4 array and calculate statistics, reshape, and resize

`reshape()` changes the shape while keeping the same number of elements. `resize()` can change the total number of elements; when the requested size is smaller, some values are removed.

```python
ucs420_Jahanvi = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array (ucs420_Jahanvi):")
print(ucs420_Jahanvi)

print("\nMean:", np.mean(ucs420_Jahanvi))
print("Median:", np.median(ucs420_Jahanvi))
print("Maximum:", np.max(ucs420_Jahanvi))
print("Minimum:", np.min(ucs420_Jahanvi))
print("Unique elements:", np.unique(ucs420_Jahanvi))

reshaped_ucs420_Jahanvi = ucs420_Jahanvi.reshape(4, 3)
print("\nReshaped array (4 rows × 3 columns):")
print(reshaped_ucs420_Jahanvi)

resized_ucs420_Jahanvi = np.resize(ucs420_Jahanvi, (2, 3))
print("\nResized array (2 rows × 3 columns):")
print(resized_ucs420_Jahanvi)
```

## Quick notes
- NumPy indexing starts at `0`, so row 1, column 2 is `arr[0, 1]`.
- `np.unique()` returns the unique values in sorted order.
- A 1-D array has no row/column orientation, so `.T` alone does not change it.
- Q5 uses `np.resize()` to produce a 2 × 3 array, retaining the first six elements in row-major order.
