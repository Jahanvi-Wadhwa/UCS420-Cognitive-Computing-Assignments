```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
df['species'] = iris.target
```

```python
print("First 5 Rows:")
print(df.head())
```

```python
print("\nShape:", df.shape)
print("\nColumns:")
print(df.columns)
```

```python
print("\nSummary Statistics:")
print(df.describe())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nMean of Sepal Length:")
print(np.mean(df['sepal length (cm)']))
```

```python
print("\nMaximum Petal Length:")
print(np.max(df['petal length (cm)']))
```

```python
print("\nMinimum Petal Width:")
print(np.min(df['petal width (cm)']))
```

```python
print("\nStandard Deviation of Sepal Width:")
print(np.std(df['sepal width (cm)']))
```
