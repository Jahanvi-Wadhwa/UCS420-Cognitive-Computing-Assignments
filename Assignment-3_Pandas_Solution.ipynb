## Import Pandas

## Q1. Create the dataset

```python
# Creating a sample dataset with 10 rows and 5 columns
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward',
             'Fiona', 'George', 'Helen', 'Ian', 'Jack'],
    'Age': [25, 30, 35, 28, 40, 22, 31, 27, 36, 29],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales',
                   'HR', 'IT', 'Marketing', 'Sales', 'IT'],
    'Salary': [45000, 60000, 65000, 52000, 70000,
               42000, 62000, 55000, 68000, 59000],
    'Experience': [2, 5, 8, 3, 12, 1, 6, 4, 10, 5]
}

df = pd.DataFrame(data)
df
```

## Q3. Navigate the DataFrame

### 1. Select rows from index 3 to 7
### 2. Select rows from index 4 to 8 and columns 2 to 4
### 3. Select all rows with column index 1 to 3

```python
# 1. Rows from index 3 to 7
print("Rows 3 to 7:")
display(df.loc[3:7])

# 2. Rows 4 to 8 and columns 2 to 4
# iloc uses integer positions, so both endpoints are included explicitly.
print("\nRows 4 to 8, columns 2 to 4:")
display(df.iloc[4:9, 2:5])

# 3. All rows with column positions 1 to 3
print("\nAll rows, columns 1 to 3:")
display(df.iloc[:, 1:4])
```

## Q5. Delete row 4 and column 3 from the CSV DataFrame

Here, row 4 and column 3 are treated as **index positions** (0-based), using `iloc` to identify them.

```python
# Delete row at index position 4
iris_modified = iris_df.drop(index=iris_df.index[4])

# Delete column at position 3
column_to_delete = iris_modified.columns[3]
iris_modified = iris_modified.drop(columns=column_to_delete)

print("Deleted row position 4 and column position 3:")
display(iris_modified.head(10))
```

## Q6. Create `employees.csv` dataset

```python
employees_data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01',
                     '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees_df = pd.DataFrame(employees_data)
employees_df.to_csv("employees.csv", index=False)

employees_df
```

### Q6(a). Shape of the DataFrame

```python
rows, columns = employees_df.shape
print("Number of rows:", rows)
print("Number of columns:", columns)
```

### Q6(b). Summary: data types and non-null counts

```python
employees_df.info()
```

### Q6(c). Descriptive statistics

```python
employees_df.describe()
```

### Q6(d). First 5 rows and last 3 rows

```python
print("First 5 rows:")
display(employees_df.head(5))

print("Last 3 rows:")
display(employees_df.tail(3))
```

### Q6(e). Required statistics

```python
average_salary = employees_df['Salary'].mean()
total_bonus = employees_df['Bonus'].sum()
youngest_age = employees_df['Age'].min()
highest_rating = employees_df['Rating'].max()

print("i. Average salary:", average_salary)
print("ii. Total bonus:", total_bonus)
print("iii. Youngest employee's age:", youngest_age)
print("iv. Highest performance rating:", highest_rating)
```

### Q6(f). Sort by Salary in descending order

```python
employees_sorted = employees_df.sort_values(by='Salary', ascending=False)
employees_sorted
```

### Q6(g). Add Performance Category

```python
def performance_category(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

employees_df['Performance_Category'] = employees_df['Rating'].apply(performance_category)
employees_df
```

### Q6(h). Identify missing values

```python
employees_df.isnull().sum()
```

### Q6(i). Rename `Employee_ID` to `ID`

```python
employees_df.rename(columns={'Employee_ID': 'ID'}, inplace=True)
employees_df
```

### Q6(j). Find employees with more than 5 years of experience and employees in IT

```python
print("i. Employees with more than 5 years of experience:")
display(employees_df[employees_df['Years_of_Experience'] > 5])

print("ii. Employees belonging to the IT department:")
display(employees_df[employees_df['Department'] == 'IT'])
```

### Q6(k). Add Tax column (10% of Salary)

```python
employees_df['Tax'] = employees_df['Salary'] * 0.10
employees_df
```

### Q6(l). Save the modified DataFrame to a new CSV file

```python
output_file = "employees_modified.csv"
employees_df.to_csv(output_file, index=False)

print("Modified DataFrame saved as:", output_file)
display(employees_df)
```
